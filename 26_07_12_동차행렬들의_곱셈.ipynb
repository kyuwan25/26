```python
#다관절로봇팔에서 dh_table을 통해 end-effector의 위치와 각도를 알 수 있는 코드를 만들었다.
import numpy as np

def DH(a,alpha,d,theta):
  ap,t = np.deg2rad(alpha), np.deg2rad(theta)
  c,s = np.cos(t),np.sin(t)
  ca,sa = np.cos(ap),np.sin(ap)
  T = np.array([[c,-s,0,a],
                  [s*ca,c*ca,-sa,-sa*d],
                  [s*sa,c*sa,ca,ca*d],
                  [0,0,0,1]])
  return T
print("i=1일때 입니다.")
a0 = float(input("a'0' : "))
alpha0 = float(input("alpha'0' : "))
d1 = float(input("d'1' : "))
theta1 = float(input("theta'1' : "))
print("i=2일때 입니다.")
a1 = float(input("a'1' : "))
alpha1 = float(input("alpha'1' : "))
d2 = float(input("d'2' : "))
theta2 = float(input("theta'2' : "))
print("i=3일때 입니다.")
a2 = float(input("a'2' : "))
alpha2 = float(input("alpha'2' : "))
d3 = float(input("d'3' : "))
theta3 = float(input("theta'3' : "))

#클로드의 조언을 바탕으로 for문을 통해서 곱셈을 구현하였다.
dh_table = [(a0,alpha0,d1,theta1),
             (a1,alpha1,d2,theta2),
             (a2,alpha2,d3,theta3)]

T_total = np.eye(4)
for params in dh_table:
    Ti = DH(*params)
    T_total = T_total @ Ti
T_result = np.round(T_total,4)
print("end-effector의 위치")
print(T_result[0:3,3])
print("end-effector의 자세")
print(T_result[0:3,0:3])
```
