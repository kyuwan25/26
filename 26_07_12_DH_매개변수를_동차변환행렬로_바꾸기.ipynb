```python
#DH parameter의 값들을 알고있을때 이를 자동으로 transformation matrices로 변환하는 코드를 만들어보겠다.
import numpy as np

def DH(a,alpha,d,theta):
  ap,t = np.deg2rad(alpha), np.deg2rad(theta)
  c,s = np.cos(t),np.sin(t)
  ca,sa = np.cos(ap),np.sin(ap)
  T = np.array([[c,-s,0,a],
                  [s*ca,c*ca,-sa,-sa*d],
                  [s*sa,c*sa,ca,ca*d],
                  [0,0,0,1]])
 #클로드의 조언으로 np.round를 넣어 부동소수점 오차를 제거했다.
  return np.round(T, 4)


a = float(input("a'i-1' : "))
alpha = float(input("alpha'i-1' : "))
d = float(input("d'i' : "))
theta = float(input("theta'i' : "))

DH(a,alpha,d,theta)

```
