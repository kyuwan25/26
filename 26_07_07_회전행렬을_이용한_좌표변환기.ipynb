```python
# 회전행렬을 이용하여 x,y,z축으로 회전하는 백터A에 대한 결과값 백터B를 구해보자
import numpy as np
A = np.array([[1],[1],[1]])
# x축을 회전축으로 하는 회전을 구하기 위한 회전행렬 R_x
theta_x = float(input("x축 회전각도를 입력하세요 : "))
theta_x = np.deg2rad(theta_x)
R_x = np.array([[1,0,0],[0,np.cos(theta_x),-np.sin(theta_x)],[0,np.sin(theta_x),np.cos(theta_x)]])
# y축을 회전축으로 하는 회전을 구하기 위한 회전행렬 R_y
theta_y = float(input("y축 회전각도를 입력하세요 : "))
theta_y = np.deg2rad(theta_y)
R_y = np.array([[np.cos(theta_y),0,np.sin(theta_y)],[0,1,0],[-np.sin(theta_y),0,np.cos(theta_y)]])
# z축을 회전축으로 하는 회전을 구하기 위한 회전행렬 R_z
theta_z = float(input("z축 회전각도를 입력하세요 : "))
theta_z = np.deg2rad(theta_z)
R_z = np.array([[np.cos(theta_z),-np.sin(theta_z),0],[np.sin(theta_z),np.cos(theta_z),0],[0,0,1]])
# 회전행렬 곱셈
B = R_x @ R_y @ R_z @ A
print(B)
#클로드의 조언을 바탕으로 기존의 외재적회전(회전행렬곱해지는 순서 반대로)에서 내재적회전으로 바꾸었다.
```

```python
#클로드가 제안한 코드 복붙함.
import numpy as np

def rotation_matrix(axis, theta_deg):
    """axis: 'x', 'y', 'z' / theta_deg: 회전각(도)"""
    t = np.deg2rad(theta_deg)
    c, s = np.cos(t), np.sin(t)
    if axis == 'x':
        return np.array([[1, 0, 0],
                          [0, c, -s],
                          [0, s,  c]])
    elif axis == 'y':
        return np.array([[ c, 0, s],
                          [ 0, 1, 0],
                          [-s, 0, c]])
    elif axis == 'z':
        return np.array([[c, -s, 0],
                          [s,  c, 0],
                          [0,  0, 1]])
    else:
        raise ValueError("axis는 'x','y','z' 중 하나여야 합니다.")

A = np.array([[1], [1], [1]])

theta_x = float(input("x축 회전각도를 입력하세요 : "))
theta_y = float(input("y축 회전각도를 입력하세요 : "))
theta_z = float(input("z축 회전각도를 입력하세요 : "))

R_x = rotation_matrix('x', theta_x)
R_y = rotation_matrix('y', theta_y)
R_z = rotation_matrix('z', theta_z)

B = R_z @ R_y @ R_x @ A
print(B)
```
